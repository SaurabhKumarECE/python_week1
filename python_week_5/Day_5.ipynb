work together correctly.

```python
import pandas as pd

from sklearn.linear_model import LinearRegression
```

```python
# Dataset

data = {
    "Experience": [1, 2, 3, 4, 5, 6],
    "Salary": [30000, 40000, 50000, 60000, 70000, 80000]
}

df = pd.DataFrame(data)

df
```

```python
# Features and target

X = df[["Experience"]]
y = df["Salary"]
```

```python
# Train model

model = LinearRegression()

model.fit(X, y)
```

```python
# Multiple test predictions

test_values = [2, 5, 8, 10, 15]

for exp in test_values:

    prediction = model.predict([[exp]])

    print(f"Experience: {exp} years")
    print(f"Predicted Salary: ₹{prediction[0]:,.2f}")
    print("-" * 40)
```
