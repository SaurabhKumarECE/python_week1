## Objective

## What is Flask?

Flask is a lightweight Python web framework used to:
- Build APIs
- Create backend services
- Connect Machine Learning models with applications

Advantages:
- Simple
- Lightweight
- Easy to learn
- Widely used

````python
import pandas as pd

from sklearn.linear_model import LinearRegression
````

````python
# Dataset

data = {
    "Experience": [1, 2, 3, 4, 5, 6],
    "Salary": [30000, 40000, 50000, 60000, 70000, 80000]
}

df = pd.DataFrame(data)

df
````

````python
# Features and target

X = df[["Experience"]]
y = df["Salary"]
````

````python
# Train model

model = LinearRegression()

model.fit(X, y)
````

````python
# Test prediction

prediction = model.predict([[9]])

print("Predicted Salary:", prediction)
````

## Flask API Workflow

User Input → Flask API → ML Model → Prediction Output
