## Objective

Finalize the capstone project problem statement and dataset.

## Problem Statement

The goal of this project is to predict house prices based on various property features.

Users will enter property information and the Machine Learning model will estimate the house price.

## Project Scope

Input Features:

- Area (sq ft)
- Bedrooms
- Bathrooms
- Floors
- Parking

Output:

- Predicted House Price

## Technologies Used

- Python
- Pandas
- Scikit-Learn
- Streamlit
- Flask
- Joblib

## Workflow

1. Dataset Collection
2. Data Preprocessing
3. Model Training
4. Hyperparameter Tuning
5. Dashboard Development
6. API Development
7. Testing
8. Deployment

## Expected Outcome

A working House Price Prediction web application capable of estimating house prices based on user inputs.
