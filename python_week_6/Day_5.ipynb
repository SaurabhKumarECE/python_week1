# Validation & Testing

## Objective

Test the House Price Prediction system with multiple house configurations.

## What is Testing?

Testing ensures that:

- Input is accepted correctly
- Predictions are generated
- Application behaves as expected

```python
import pandas as pd
import joblib
```

```python
# Load model

model = joblib.load(
    "best_house_price_model.pkl"
)

print("Model Loaded Successfully")
```

```python
# Test Case 1

house_1 = pd.DataFrame(
    {
        "Area": [1500],
        "Bedrooms": [3],
        "Bathrooms": [2],
        "Floors": [1],
        "Parking": [1]
    }
)

prediction = model.predict(house_1)

print("Test Case 1")
print("Predicted Price:", prediction[0])
```

```python
# Test Case 2

house_2 = pd.DataFrame(
    {
        "Area": [2500],
        "Bedrooms": [5],
        "Bathrooms": [4],
        "Floors": [3],
        "Parking": [3]
    }
)

prediction = model.predict(house_2)

print("Test Case 2")
print("Predicted Price:", prediction[0])
```

```python
# Test Case 3

house_3 = pd.DataFrame(
    {
        "Area": [3500],
        "Bedrooms": [6],
        "Bathrooms": [6],
        "Floors": [3],
        "Parking": [4]
    }
)

prediction = model.predict(house_3)

print("Test Case 3")
print("Predicted Price:", prediction[0])
```

```python
# Multiple Testing

test_data = pd.DataFrame(
    {
        "Area": [1200, 1800, 2700],
        "Bedrooms": [2, 3, 5],
        "Bathrooms": [2, 3, 5],
        "Floors": [1, 2, 3],
        "Parking": [1, 2, 3]
    }
)

predictions = model.predict(test_data)

for i, price in enumerate(predictions):
    print(f"House {i+1}: ₹{price:,.0f}")
```

## Test Results

The model was tested using multiple house configurations.

All predictions were generated successfully.

- Verified prediction outputs
